# 01 环境检查

把下面每个单元格跑一遍（选中单元格，按 Shift+Enter）。全部打 ✅ 就说明环境没问题，可以往下学了。

如果有 ❌，回到仓库根目录的终端里跑一次：

```
pip install -r requirements.txt
```

````python
import sys, platform
print("Python", sys.version.split()[0], "on", platform.system())
````

````python
libs = ["numpy", "pandas", "matplotlib", "sklearn", "jupyterlab"]

for name in libs:
    try:
        mod = __import__(name)
        print("✅", name, getattr(mod, "__version__", ""))
    except ImportError:
        print("❌", name, "还没装 —— 在终端跑 pip install -r requirements.txt")
````
